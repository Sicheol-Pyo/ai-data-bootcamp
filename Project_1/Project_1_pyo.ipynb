```python
# 점포_자치구 전체 점포 수 대비 폐업률 상관분석

group_stats = 점포_자치구_EDA.groupby("자치구_코드_명").agg(
    평균_점포수=("전체_점포_수", "mean"),
    평균_폐업률=("폐업_률", "mean"),
    폐업률_표준편차=("폐업_률", "std"),
).reset_index()

# 상관관계 (선형: Pearson, 순위 기반: Spearman)
corr_mean, p_mean = stats.pearsonr(group_stats["평균_점포수"], group_stats["평균_폐업률"])
corr_std, p_std = stats.pearsonr(group_stats["평균_점포수"], group_stats["폐업률_표준편차"])
spear_mean, sp_p_mean = stats.spearmanr(group_stats["평균_점포수"], group_stats["평균_폐업률"])
spear_std, sp_p_std = stats.spearmanr(group_stats["평균_점포수"], group_stats["폐업률_표준편차"])

print(f"점포수 vs 평균 폐업률   : Pearson r={corr_mean:.3f} (p={p_mean:.4g}), Spearman r={spear_mean:.3f} (p={sp_p_mean:.4g})")
print(f"점포수 vs 폐업률 변동성  : Pearson r={corr_std:.3f} (p={p_std:.4g}), Spearman r={spear_std:.3f} (p={sp_p_std:.4g})")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].scatter(group_stats["평균_점포수"], group_stats["평균_폐업률"], color="#2a78d6", s=60)
for _, row in group_stats.iterrows():
    axes[0].annotate(row["자치구_코드_명"], (row["평균_점포수"], row["평균_폐업률"]), fontsize=7, alpha=0.7)
axes[0].set_xlabel("평균 전체_점포_수")
axes[0].set_ylabel("평균 폐업_률 (%)")
axes[0].set_title(f"점포 수 vs 평균 폐업률 (r={corr_mean:.2f})")

axes[1].scatter(group_stats["평균_점포수"], group_stats["폐업률_표준편차"], color="#eb6834", s=60)
for _, row in group_stats.iterrows():
    axes[1].annotate(row["자치구_코드_명"], (row["평균_점포수"], row["폐업률_표준편차"]), fontsize=7, alpha=0.7)
axes[1].set_xlabel("평균 전체_점포_수")
axes[1].set_ylabel("폐업_률 표준편차")
axes[1].set_title(f"점포 수 vs 폐업률 변동성 (r={corr_std:.2f})")

plt.tight_layout()
plt.show()
```

```python
# 업종별로 상권유형(상권_구분_코드_명) 간 폐업률 차이를 Kruskal-Wallis로 검정

업종_목록 = 점포_상권_EDA["서비스_업종_코드_명"].unique()
alpha = 0.05
n_tests = len(업종_목록)
bonferroni_alpha = alpha / n_tests

결과 = []
for 업종명 in 업종_목록:
    sub = 점포_상권_EDA[점포_상권_EDA["서비스_업종_코드_명"] == 업종명]
    groups = [g["폐업_률"].values for _, g in sub.groupby("상권_구분_코드_명")]
    h_stat, p_value = stats.kruskal(*groups)
    결과.append({
        "업종": 업종명,
        "H": round(h_stat, 3),
        "p_value": p_value,
        "유의_Bonferroni적용": p_value < bonferroni_alpha,
    })

결과_df = pd.DataFrame(결과).sort_values("p_value")
print(f"검정 횟수: {n_tests}개, Bonferroni 보정 기준 alpha = {bonferroni_alpha:.5f} (0.05 ÷ {n_tests})")
print(결과_df)
```

```python
# 업종별 Kruskal-Wallis 효과크기(epsilon-squared) 계산

# epsilon^2 = (H - k + 1) / (n - k)   [k: 그룹 수, n: 전체 표본 수]
효과크기_결과 = []
for 업종명 in 업종_목록:
    sub = 점포_상권_EDA[점포_상권_EDA["서비스_업종_코드_명"] == 업종명]
    groups = [g["폐업_률"].values for _, g in sub.groupby("상권_구분_코드_명")]
    h_stat, p_value = stats.kruskal(*groups)
    k = len(groups)
    n = sum(len(g) for g in groups)
    epsilon_sq = (h_stat - k + 1) / (n - k)
    효과크기_결과.append({"업종": 업종명, "H": round(h_stat,1), "n": n, "epsilon_squared": round(epsilon_sq, 4)})

효과크기_df = pd.DataFrame(효과크기_결과).sort_values("epsilon_squared", ascending=False)
print(효과크기_df)
```

```python
# 업종별로 자치구(자치구_코드_명) 간 폐업률 차이를 Kruskal-Wallis로 검정
# (다중비교 문제 보정을 위해 Bonferroni 방법 적용)
업종_목록 = 점포_자치구_EDA["서비스_업종_코드_명"].unique()
alpha = 0.05
n_tests = len(업종_목록)
bonferroni_alpha = alpha / n_tests

결과 = []
for 업종명 in 업종_목록:
    sub = 점포_자치구_EDA[점포_자치구_EDA["서비스_업종_코드_명"] == 업종명]
    groups = [g["폐업_률"].values for _, g in sub.groupby("자치구_코드_명")]
    h_stat, p_value = stats.kruskal(*groups)
    k = len(groups)
    n = sum(len(g) for g in groups)
    epsilon_sq = (h_stat - k + 1) / (n - k)
    결과.append({
        "업종": 업종명,
        "H": round(h_stat, 3),
        "n": n,
        "p_value": p_value,
        "유의_Bonferroni적용": p_value < bonferroni_alpha,
        "epsilon_squared": round(epsilon_sq, 4),
    })

결과_df = pd.DataFrame(결과).sort_values("epsilon_squared", ascending=False)
print(f"검정 횟수: {n_tests}개, Bonferroni 보정 기준 alpha = {bonferroni_alpha:.5f} (0.05 ÷ {n_tests})")
print(결과_df)
```

```python
# 모델별 홀드아웃 성능(MAE, RMSE, R²)을 한 번에 계산해주는 공통 평가 함수

def 평가(model, test_df, target_col="폐업_률_다음분기"):
    pred = model.predict(test_df)
    mae = mean_absolute_error(test_df[target_col], pred)
    rmse = np.sqrt(mean_squared_error(test_df[target_col], pred))
    r2 = r2_score(test_df[target_col], pred)
    return mae, rmse, r2
```

```python
# 데이터 준비: 자치구x업종 그룹별로 "다음 분기 폐업_률"을 타겟으로 만들고, train(20231~20253)/test(20254→20261) 분리
점포_자치구_예측 = 점포_자치구_final.sort_values(["자치구_코드_명","서비스_업종_코드_명","기준_년분기_코드"]).copy()
점포_자치구_예측["폐업_률_다음분기"] = 점포_자치구_예측.groupby(["자치구_코드_명","서비스_업종_코드_명"])["폐업_률"].shift(-1)
점포_자치구_예측["분기_순번"] = 점포_자치구_예측["기준_년분기_코드"] % 10
점포_자치구_예측["프랜차이즈_비율"] = 점포_자치구_예측["프랜차이즈_점포_수"] / 점포_자치구_예측["전체_점포_수"]
점포_자치구_예측["log_전체_점포_수"] = np.log1p(점포_자치구_예측["전체_점포_수"])

점포_자치구_예측_valid = 점포_자치구_예측.dropna(subset=["폐업_률_다음분기"]).copy()
train_자치구예측 = 점포_자치구_예측_valid[점포_자치구_예측_valid["기준_년분기_코드"] != 20254].copy()
test_자치구예측  = 점포_자치구_예측_valid[점포_자치구_예측_valid["기준_년분기_코드"] == 20254].copy()
```

```python
# 모델 1: naive 기준선 - 이번 분기 값을 다음 분기 예측치로 그대로 사용 (계절성 미반영)
naive_pred = test_자치구예측["폐업_률"]
naive_mae = mean_absolute_error(test_자치구예측["폐업_률_다음분기"], naive_pred)
naive_rmse = np.sqrt(mean_squared_error(test_자치구예측["폐업_률_다음분기"], naive_pred))
naive_r2 = r2_score(test_자치구예측["폐업_률_다음분기"], naive_pred)
print(f"[모델1: naive] MAE={naive_mae:.3f}, RMSE={naive_rmse:.3f}, R²={naive_r2:.4f}")
```

```python
# 모델 2: pooled OLS (채택 모델) - 자치구/업종 더미 + 개업률/점포수/프랜차이즈비율/계절성 반영
model_next2 = smf.ols(
    "폐업_률_다음분기 ~ 폐업_률 + 개업_율 + log_전체_점포_수 + 프랜차이즈_비율 "
    "+ C(자치구_코드_명) + C(서비스_업종_코드_명) + C(분기_순번)",
    data=train_자치구예측
).fit(cov_type="HC3")

mae2, rmse2, r2_2 = 평가(model_next2, test_자치구예측)
print(f"[모델2: pooled OLS] MAE={mae2:.3f}, RMSE={rmse2:.3f}, R²={r2_2:.4f}")
```

```python
# 모델 3: 그룹별 자체 평균 - 각 (자치구,업종) 그룹의 학습기간 평균 폐업률을 그대로 예측치로 사용
group_avg = train_자치구예측.groupby(["자치구_코드_명","서비스_업종_코드_명"])["폐업_률"].mean().reset_index()
group_avg.columns = ["자치구_코드_명","서비스_업종_코드_명","그룹평균_예측"]
test_group_avg = test_자치구예측.merge(group_avg, on=["자치구_코드_명","서비스_업종_코드_명"], how="left")

mae_g = mean_absolute_error(test_group_avg["폐업_률_다음분기"], test_group_avg["그룹평균_예측"])
rmse_g = np.sqrt(mean_squared_error(test_group_avg["폐업_률_다음분기"], test_group_avg["그룹평균_예측"]))
r2_g = r2_score(test_group_avg["폐업_률_다음분기"], test_group_avg["그룹평균_예측"])
print(f"[모델3: 그룹별 자체 평균] MAE={mae_g:.3f}, RMSE={rmse_g:.3f}, R²={r2_g:.4f}")
```

```python
# 모델 4: 그룹별 개별 추세선 - 각 (자치구,업종)마다 자기 자신의 시계열만으로 선형 추세선을 학습해 다음 시점 예측
train_자치구예측["시간인덱스"] = train_자치구예측.groupby(["자치구_코드_명","서비스_업종_코드_명"]).cumcount() + 1

results = []
for (자치구, 업종), group in train_자치구예측.groupby(["자치구_코드_명","서비스_업종_코드_명"]):
    if len(group) < 3:
        continue
    m = smf.ols("폐업_률 ~ 시간인덱스", data=group).fit()
    다음_시간인덱스 = group["시간인덱스"].max() + 1
    예측값 = max(0, m.predict(pd.DataFrame({"시간인덱스":[다음_시간인덱스]}))[0])
    results.append({"자치구_코드_명":자치구, "서비스_업종_코드_명":업종, "개별추세_예측": 예측값})

개별예측_df = pd.DataFrame(results)
test_개별추세 = test_자치구예측.merge(개별예측_df, on=["자치구_코드_명","서비스_업종_코드_명"], how="left")

mae_t = mean_absolute_error(test_개별추세["폐업_률_다음분기"], test_개별추세["개별추세_예측"])
rmse_t = np.sqrt(mean_squared_error(test_개별추세["폐업_률_다음분기"], test_개별추세["개별추세_예측"]))
r2_t = r2_score(test_개별추세["폐업_률_다음분기"], test_개별추세["개별추세_예측"])
print(f"[모델4: 그룹별 개별 추세선] MAE={mae_t:.3f}, RMSE={rmse_t:.3f}, R²={r2_t:.4f}")
```

```python
# 모델 5: XGBoost - 부스팅 기반 트리 모델 (범주형 변수는 LabelEncoder로 숫자 인코딩)
import xgboost as xgb
from sklearn.preprocessing import LabelEncoder

train_ml, test_ml = train_자치구예측.copy(), test_자치구예측.copy()
for col in ["자치구_코드_명", "서비스_업종_코드_명"]:
    le = LabelEncoder()
    le.fit(pd.concat([train_ml[col], test_ml[col]]))
    train_ml[col+"_enc"] = le.transform(train_ml[col])
    test_ml[col+"_enc"] = le.transform(test_ml[col])

features = ["폐업_률","개업_율","log_전체_점포_수","프랜차이즈_비율","분기_순번","자치구_코드_명_enc","서비스_업종_코드_명_enc"]
X_train, y_train = train_ml[features], train_ml["폐업_률_다음분기"]
X_test, y_test = test_ml[features], test_ml["폐업_률_다음분기"]

model_xgb = xgb.XGBRegressor(n_estimators=200, max_depth=3, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8, random_state=42)
model_xgb.fit(X_train, y_train)
pred_xgb = model_xgb.predict(X_test)

mae_x = mean_absolute_error(y_test, pred_xgb)
rmse_x = np.sqrt(mean_squared_error(y_test, pred_xgb))
r2_x = r2_score(y_test, pred_xgb)
print(f"[모델5: XGBoost] MAE={mae_x:.3f}, RMSE={rmse_x:.3f}, R²={r2_x:.4f}")
```

```python
# 모델 6: RandomForest(배깅) - 여러 트리를 독립적으로 학습해 평균내는 방식 (모델5의 X_train/y_train 재사용)
from sklearn.ensemble import RandomForestRegressor

model_rf = RandomForestRegressor(n_estimators=300, max_depth=5, min_samples_leaf=10, random_state=42)
model_rf.fit(X_train, y_train)
pred_rf = model_rf.predict(X_test)

mae_r = mean_absolute_error(y_test, pred_rf)
rmse_r = np.sqrt(mean_squared_error(y_test, pred_rf))
r2_r = r2_score(y_test, pred_rf)
print(f"[모델6: RandomForest] MAE={mae_r:.3f}, RMSE={rmse_r:.3f}, R²={r2_r:.4f}")
```

```python
# 모델 7: Walk-forward 검증 - 학습 구간을 한 분기씩 늘려가며 모델2와 동일한 구조로 8개 시점을 반복 검증
quarters_순서 = sorted(점포_자치구_예측_valid["기준_년분기_코드"].unique())
results_wf = []
for i in range(4, len(quarters_순서)):
    train_분기, test_분기 = quarters_순서[:i], quarters_순서[i]
    train_fold = 점포_자치구_예측_valid[점포_자치구_예측_valid["기준_년분기_코드"].isin(train_분기)]
    test_fold = 점포_자치구_예측_valid[점포_자치구_예측_valid["기준_년분기_코드"] == test_분기]

    model_fold = smf.ols(
        "폐업_률_다음분기 ~ 폐업_률 + 개업_율 + log_전체_점포_수 + 프랜차이즈_비율 "
        "+ C(자치구_코드_명) + C(서비스_업종_코드_명) + C(분기_순번)",
        data=train_fold
    ).fit()

    mae_f, rmse_f, r2_f = 평가(model_fold, test_fold)
    results_wf.append({"기준_분기": test_분기, "학습분기수": len(train_분기), "MAE": round(mae_f,3), "RMSE": round(rmse_f,3), "R2": round(r2_f,4)})

결과_walkforward = pd.DataFrame(results_wf)
print("[모델7: Walk-forward 검증]")
print(결과_walkforward)
```

```python
# 모델 13: 상권x업종 개별 추세선 - 상권_코드 단위에서도 자기 자신의 시계열만으로 추세선 예측 (numpy.polyfit로 속도 개선)
train_상권예측["시간인덱스"] = train_상권예측.groupby(["상권_코드","서비스_업종_코드_명"]).cumcount() + 1

def fit_predict(group):
    if len(group) < 3:
        return np.nan
    x, y = group["시간인덱스"].values, group["폐업_률"].values
    slope, intercept = np.polyfit(x, y, 1)
    예측값 = slope*(x.max()+1) + intercept
    return max(0, min(100, 예측값))

개별예측_상권_df = train_상권예측.groupby(["상권_코드","서비스_업종_코드_명"]).apply(fit_predict).reset_index()
개별예측_상권_df.columns = ["상권_코드","서비스_업종_코드_명","개별추세_예측"]

test_개별추세_상권 = test_상권예측.merge(개별예측_상권_df, on=["상권_코드","서비스_업종_코드_명"], how="left").dropna(subset=["개별추세_예측"])

mae_ts = mean_absolute_error(test_개별추세_상권["폐업_률_다음분기"], test_개별추세_상권["개별추세_예측"])
rmse_ts = np.sqrt(mean_squared_error(test_개별추세_상권["폐업_률_다음분기"], test_개별추세_상권["개별추세_예측"]))
r2_ts = r2_score(test_개별추세_상권["폐업_률_다음분기"], test_개별추세_상권["개별추세_예측"])
print(f"[모델13: 상권x업종 개별 추세선] MAE={mae_ts:.3f}, RMSE={rmse_ts:.3f}, R²={r2_ts:.4f}")
```

```python
def 가중평가(model, test_df, target_col="폐업_률_다음분기", weight_col="전체_점포_수"):
    pred = model.predict(test_df)
    err = pred - test_df[target_col]
    w = test_df[weight_col]

    가중_mae = np.average(err.abs(), weights=w)
    가중_rmse = np.sqrt(np.average(err**2, weights=w))
    return 가중_mae, 가중_rmse

mae_w, rmse_w = 가중평가(model_상권유형, test_상권예측)
print(f"가중 MAE={mae_w:.3f}, 가중 RMSE={rmse_w:.3f}")
```
