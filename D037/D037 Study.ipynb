```python
# 미션 1 · 준비 — 읽고 실행만 합니다. 이 셀이 미션 1 에 쓸 변수를 직접 만듭니다
from sklearn.datasets import fetch_openml

# freMTPL2freq — 프랑스 자동차보험 계약별 사고 빈도 (678,013 x 12). OpenML 에서 바로 내려받습니다
df = fetch_openml(data_id=41214, as_frame=True).frame
df["IDpol"] = df["IDpol"].astype(int)
for c in ["Area", "VehBrand", "VehGas", "Region"]:
    df[c] = df[c].astype(str).str.strip("'").astype("category")

# 4일 동안 고정해 온 전처리 2가지
df["ClaimNb"] = df["ClaimNb"].astype(int).clip(upper=4)    # 사고 5건 이상은 극소수라 상한 4 로 절사
df["Exposure"] = df["Exposure"].clip(upper=1.0)            # 노출은 1년 상한

LAM_ALL = df["ClaimNb"].sum() / df["Exposure"].sum()       # 쓸 변수 1 — 전체 계약(회사 전체) 사고율 (건/년)
B, SEED = 1000, 47                                         # 쓸 변수 2 — 재표본 수와 난수 시드

SEGMENTS = {f"Area {a}": (df["Area"] == a) for a in list("ABCDEF")}   # 쓸 변수 3 — 세그먼트 이름 -> 불리언 마스크
SEGMENTS["Region R43"] = (df["Region"] == "R43")              # 계약 수가 가장 적은 지역
SEGMENTS["Region R42"] = (df["Region"] == "R42")              # 그다음으로 계약 수가 적은 지역

for seg_name, mask in SEGMENTS.items():
    d = df[mask]
    print(f"{seg_name:12s} 계약 {len(d):7,d}건  사고 {int(d['ClaimNb'].sum()):5,d}건  "
          f"노출 {d['Exposure'].sum():10,.1f}년  "
          f"점추정 {d['ClaimNb'].sum() / d['Exposure'].sum():.6f} 건/년")

print(f"\n전체 계약 사고율 LAM_ALL = {LAM_ALL:.6f} 건/년 · 재표본 수 B = {B} · 난수 시드 SEED = {SEED}")
print("쓸 수 있는 이름 : df (전처리 끝난 계약 표) · SEGMENTS (세그먼트 이름 -> 마스크 8개) · "
      "B (재표본 수) · SEED (난수 시드) · LAM_ALL (전체 계약 사고율, 건/년)")
```

```python
# 미션 1 — 프롬프트   # ai: prompt
# PROMPT 를 쓰고 → 도우미에 "위 PROMPT 대로 코드를 써 줘" → 「이 셀에 넣기」 → ▶
PROMPT = """
df, SEGMENTS, B, SEED, LAM_ALL은 그대로 만들걸 사용해
1. Region R43의 95% 구간의 하한: 단위 건/년, 소수점 아래 5자리, 재표본 사고율을 정렬해 아래 2.5% 절단한 값
2. Region R43의 95% 구간의 상한: 단위 건/년, 소수점 아래 5자리, 재표본 사고율을 정렬해 위 2.5% 절단한 값
3. 표 하나: 세그먼트 8개마다 세그먼트, n, claims, exposure, lam_hat, lo95, hi95, se, 폭, 상대폭
4. 그래프
가로막대: 세그먼트마다 하한에서 상한까지 하나씩 모두 8개
마커: 세그먼트마다 점 추정 위치에 하나씩
세로선: LAM_ALL 위치에 세로축 전체를 가로지르는 선 하나
가로축 이름: 연간 사고율(건/년)
세로축 이름: 요율표의 세그먼트
"""
# --- 여기부터 AI 코드 ---
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(SEED)
rows = []

for seg_name, mask in SEGMENTS.items():
    sub = df[mask]
    claims = sub['ClaimNb'].to_numpy()
    exposure = sub['Exposure'].to_numpy()
    n = len(sub)
    lam_hat = claims.sum() / exposure.sum()

    boot_lams = np.empty(B)
    for b in range(B):
        idx = rng.integers(0, n, size=n)
        boot_lams[b] = claims[idx].sum() / exposure[idx].sum()

    boot_lams.sort()
    lo95 = np.percentile(boot_lams, 2.5)
    hi95 = np.percentile(boot_lams, 97.5)
    se = boot_lams.std(ddof=1)
    width = hi95 - lo95
    rel_width = width / lam_hat

    rows.append({
        '세그먼트': seg_name, 'n': n,
        'claims': claims.sum(), 'exposure': exposure.sum(),
        'lam_hat': lam_hat, 'lo95': lo95, 'hi95': hi95,
        'se': se, '폭': width, '상대폭': rel_width,
    })

result = pd.DataFrame(rows)

r43 = result.loc[result['세그먼트'] == 'Region R43'].iloc[0]
print(f"R43 95% 하한: {r43['lo95']:.5f} 건/년")
print(f"R43 95% 상한: {r43['hi95']:.5f} 건/년")
print(result)

fig, ax = plt.subplots(figsize=(7, 5))
y_pos = np.arange(len(result))
ax.hlines(y_pos, result['lo95'], result['hi95'], color='steelblue', lw=3)
ax.scatter(result['lam_hat'], y_pos, color='steelblue', zorder=3, marker='s')
ax.axvline(LAM_ALL, color='red', linestyle='--', label='전체 사고율')
ax.set_yticks(y_pos)
ax.set_yticklabels(result['세그먼트'])
ax.set_xlabel('연간 사고율(건/년)')
ax.set_ylabel('요율표의 세그먼트')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# 미션 2 — 프롬프트 (함정)   # ai: prompt
# 이 미션은 함정입니다 — 아래는 전임 담당자가 남긴 코드라 돌아가지만 답이 틀렸습니다.
# 어디를 어떻게 고칠지 PROMPT 에 쓰면 튜터가 「AI 코드 시작 행」 아래를 다시 씁니다.
PROMPT = """
df, Xdesign, Xz, COLS, y_rate, w_exp, ALPHAS, SEED는 그대로 사용
나는 아래 있는 전임 담당자 코드가 alpha 격자를 6개만 설정해서 답이 틀린 것으로 추측했어
실제로 점검해서 알려주고
1. 교차 검증 이탈도가 가장 낮은 지점의 이탈도: 소수점 6자리
2. 벌점이 거의 없는 alpha에서의 표준화 계수 40개의 크기 합: 소수점 4자리
3. 그래프 한장(패널 2개)
왼쪽 계수 경로: 가로축은 정규화 강도 alpha(로그 눈금), 세로축은 표준화 계수인 선 그래프
선: COLS 40개에 하나씩, 그 가운데 BonusMalus, VehBrand_B12는 선 색을 다르게 적용
가로축 이름: 정규화 강도 alpha, 세로축 이름: 표준화 계수

오른쪽 alpha별 이탈도: 가로축은 alpha 6개를 배치한 범주형 축, 세로축은 교차 검증 포아송 이탈도인 막대그래프
막대: alpha 하나에 하나씩 총 6개, 0에서 시작하고 폴드간 +-1 표준오차의 오차막대도 표시
가로축이름: 정규화 강도 alpha, 세로축 이름: 교차검증 포아송 이탈도
"""
# --- 여기부터 AI 코드 ---
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import PoissonRegressor
from sklearn.model_selection import KFold
from sklearn.metrics import mean_poisson_deviance

Xz_arr = Xz
y_arr = np.asarray(y_rate)
w_arr = np.asarray(w_exp)

kf = KFold(n_splits=5, shuffle=True, random_state=SEED)
cv_dev = {a: [] for a in ALPHAS}

for train_idx, test_idx in kf.split(Xz_arr):
    X_train, X_test = Xz_arr[train_idx], Xz_arr[test_idx]
    y_train, y_test = y_arr[train_idx], y_arr[test_idx]
    w_train, w_test = w_arr[train_idx], w_arr[test_idx]
    for a in ALPHAS:
        fit = PoissonRegressor(alpha=a, max_iter=300, tol=1e-6).fit(
            X_train, y_train, sample_weight=w_train)
        pred = fit.predict(X_test)
        dev = mean_poisson_deviance(y_test, pred, sample_weight=w_test)
        cv_dev[a].append(dev)

cv_mean = {a: np.mean(cv_dev[a]) for a in ALPHAS}
cv_se = {a: np.std(cv_dev[a], ddof=1) / np.sqrt(5) for a in ALPHAS}

best_alpha = min(cv_mean, key=cv_mean.get)
best_dev = cv_mean[best_alpha]
print(f"교차검증 이탈도가 가장 낮은 지점의 이탈도 = {best_dev:.6f}")

coef_path = {}
for a in ALPHAS:
    fit = PoissonRegressor(alpha=a, max_iter=300, tol=1e-6).fit(
        Xz_arr, y_arr, sample_weight=w_arr)
    coef_path[a] = fit.coef_
coef_path_tbl = pd.DataFrame(coef_path, index=COLS)

alpha_min = min(ALPHAS)
coef_sum = coef_path_tbl[alpha_min].abs().sum()
print(f"벌점이 거의 없는 alpha에서의 표준화 계수 40개의 크기 합 (alpha = {alpha_min:g}) = {coef_sum:.4f}")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

ax = axes[0]
highlight = {'BonusMalus', 'VehBrand_B12'}
for col in COLS:
    if col in highlight:
        ax.plot(ALPHAS, coef_path_tbl.loc[col], label=col, linewidth=2.5)
    else:
        ax.plot(ALPHAS, coef_path_tbl.loc[col], color='lightgray', linewidth=0.8)
ax.set_xscale('log')
ax.set_xlabel('정규화 강도 alpha')
ax.set_ylabel('표준화 계수')
ax.legend()

ax = axes[1]
alphas_str = [f"{a:g}" for a in ALPHAS]
means = [cv_mean[a] for a in ALPHAS]
ses = [cv_se[a] for a in ALPHAS]
ax.bar(alphas_str, means, yerr=ses, capsize=4)
ax.set_ylim(bottom=0)
ax.set_xlabel('정규화 강도 alpha')
ax.set_ylabel('교차검증 포아송 이탈도')

plt.tight_layout()
plt.show()
```

1. 설계행렬을 잘못된 것을 썼다 전임 코드는 Xdesign.to_numpy() (원래 단위, 표준화 안 됨)로 학습했습니다. 정규화(패널티)는 계수 크기를 직접 누르는 방식이라, 열마다 척도(scale)가 다르면 같은 강도의 패널티라도 열마다 실제로 받는 영향이 달라집니다. 예를 들어 BonusMalus처럼 값 범위가 큰 열과 0/1 더미 열은 원래 단위에서 같은 alpha를 적용해도 공평하게 눌리지 않아요. 그래서 요구사항이 "표준화 계수"였던 거고, 수정 코드는 Xz(평균 0·표준편차 1)로 바꿔 학습했습니다.

2. 교차검증이 아예 없었다 전임 코드는 Xdesign 전체로 한 번만 .fit()했을 뿐, 훈련/검증 분할도 없고 이탈도를 계산·출력하는 부분도 없었습니다. 그래서 채점이 찾는 문자열("교차검증 이탈도가 가장 낮은 지점의 이탈도 = ...")이 화면에 아예 없었던 거예요. 수정 코드는 KFold(n_splits=5, shuffle=True, random_state=SEED)로 폴드를 한 번만 나누고, 그 분할을 alpha 6개 전부에 똑같이 재사용해 mean_poisson_deviance로 이탈도를 구했습니다.

3. 그래프가 없었다 전임 코드는 계수를 alpha별로 모으기만 했지 시각화가 없었습니다. 수정 코드는 왼쪽 패널에 alpha별 계수 경로(로그축, BonusMalus·VehBrand_B12 강조), 오른쪽 패널에 alpha별 CV 이탈도 막대(오차막대 포함)를 추가했습니다.

```python
# 미션 3 · 준비 — 읽고 실행만 합니다. 이 셀이 미션 3 에 쓸 변수를 직접 만듭니다
if "df" not in globals():                                       # 이 미션만 따로 실행해도 단독으로 동작합니다
    from sklearn.datasets import fetch_openml                    # (앞 미션을 이미 돌렸으면 통째로 건너뜁니다)
    df = fetch_openml(data_id=41214, as_frame=True).frame
    df["IDpol"] = df["IDpol"].astype(int)
    for c in ["Area", "VehBrand", "VehGas", "Region"]:
        df[c] = df[c].astype(str).str.strip("'").astype("category")
    df["ClaimNb"] = df["ClaimNb"].astype(int).clip(upper=4)      # 4일 동안 고정해 온 전처리 2가지
    df["Exposure"] = df["Exposure"].clip(upper=1.0)

SEED = 47                                                       # 쓸 변수 1 — 분할 난수 시드
FOLDS = list(KFold(5, shuffle=True, random_state=SEED).split(df))   # 쓸 변수 2 — (학습, 검증) 인덱스 5쌍
y_rate = (df["ClaimNb"] / df["Exposure"]).to_numpy()            # 쓸 변수 3 — 빈도 타깃 (건/년)
w_exp = df["Exposure"].to_numpy()                               # 쓸 변수 4 — 노출 가중 (년)
LAM_ALL = float(df["ClaimNb"].sum() / df["Exposure"].sum())     # 쓸 변수 5 — 전체 계약 사고율 (건/년)

FORMULAS = {                                                    # 쓸 변수 6 — 모형 이름 -> formula 문자열
    "null": "ClaimNb ~ 1",
    "core": "ClaimNb ~ BonusMalus + DrivAge + C(Area) + C(VehGas)",
    "full": "ClaimNb ~ BonusMalus + DrivAge + C(Area) + C(VehGas)"
            " + C(VehBrand) + C(Region) + np.log(Density)"}
print("어제 정한 후보 셋을 이 셀이 그대로 다시 적습니다\n")

for i, (tr, te) in enumerate(FOLDS):
    print(f"폴드 {i + 1} : 학습 {len(tr):,}건 · 검증 {len(te):,}건 · 겹침 {len(set(tr) & set(te)):,}건")
print("\n검증할 후보 셋")
for name, formula in FORMULAS.items():
    print(f"  {name:5s} : {formula}")
print(f"\n이탈도 기준선 — 전체 계약에 같은 요율 {LAM_ALL:.6f} 건/년 하나를 매겼을 때 : "
      f"{mean_poisson_deviance(y_rate, np.full(len(df), LAM_ALL), sample_weight=w_exp):.6f}"
      "  (단위 없는 점수라 이 값과 견줘 읽습니다)")
print("쓸 수 있는 이름 : df (전처리 끝난 계약 표) · FOLDS (고정 분할 5쌍) · "
      "FORMULAS (후보 셋의 formula) · y_rate (빈도 타깃, 건/년) · w_exp (노출 가중, 년) · "
      "LAM_ALL (전체 계약 사고율, 건/년)")
```

```python
# 미션 3 — 프롬프트   # ai: prompt
# PROMPT 를 쓰고 → 도우미에 "위 PROMPT 대로 코드를 써 줘" → 「이 셀에 넣기」 → ▶
PROMPT = """
df, FOLDS, FORMULAS, y_rate, w_exp, LAM_ALL은 그대로 사용해
1. 확장 모형(full)의 폴드 평균 이탈도: 소수점 6자리, 폴드 5개 점수를 평균한 값
2. core-full 짝지은 차이의 평균/표준오차: 단위 배(표준오차의 배수), 소수점 아래 2자리
같은 폴드에서 core 의 이탈도에서 full 의 이탈도를 뺀 차이 5개를 만들고, 그 평균을 그 5개로 산출한 표준오차로 나눈 값
3. 출력할 3개 행:후보 3개마다 폴드 평균, 폴드 간 표준오차, 폴드별 점수 5개 한행씩
4. 그래프 (패널 2개)
왼쪽 후보별 폴드 점수: 가로축은 후보 3개 배치한 범주형, 세로축은 교차검증 포아송 이탈도인 산점도
점: 후보마다 폴드별 이탈도 5개, 평균 위치에 짧은 가로선 하나
가로축 이름: 후보 모형, 세로축 이름: 교차검증 포아송 이탈도

오른쪽 폴드별 짝지은 차이: 가로축은 폴드 1~5, 세로축은 core에서 full을 뺀 이탈도 차이인 막대그래프
막대: 폴드 하나에 하나씩 5개
가로선: 0위치에 실선하나, 차이의 평균 위치에 점선 하나, 2개 모두 가로축 전체를 가로지름
가로축 이름: 폴드, 세로축 이름: core-full 이탈도 차이

"""
# --- 여기부터 AI 코드 ---
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from sklearn.metrics import mean_poisson_deviance
import matplotlib.pyplot as plt

y_arr = np.asarray(y_rate)
w_arr = np.asarray(w_exp)

fold_scores = {name: [] for name in FORMULAS}

for train_idx, val_idx in FOLDS:
    train_df = df.iloc[train_idx]
    val_df = df.iloc[val_idx]
    w_train = w_arr[train_idx]
    w_val = w_arr[val_idx]
    y_val = y_arr[val_idx]

    for name, formula in FORMULAS.items():
        model = smf.glm(formula=formula, data=train_df,
                         family=sm.families.Poisson(),
                         freq_weights=w_train).fit()
        pred = model.predict(val_df)
        dev = mean_poisson_deviance(y_val, pred, sample_weight=w_val)
        fold_scores[name].append(dev)

rows = []
for name in FORMULAS:
    scores = np.array(fold_scores[name])
    row = {'후보': name,
           '폴드평균': scores.mean(),
           '폴드간표준오차': scores.std(ddof=1) / np.sqrt(len(scores))}
    for i, s in enumerate(scores, start=1):
        row[f'폴드{i}'] = s
    rows.append(row)
result = pd.DataFrame(rows)
print(result)

full_mean = result.loc[result['후보'] == 'full', '폴드평균'].iloc[0]
print(f"확장 모형(full)의 폴드 평균 이탈도 = {full_mean:.6f}")

diff = np.array(fold_scores['core']) - np.array(fold_scores['full'])
diff_mean = diff.mean()
diff_se = diff.std(ddof=1) / np.sqrt(len(diff))
t_stat = diff_mean / diff_se
print(f"core-full 짝지은 차이의 평균/표준오차 = {t_stat:.2f}배")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

ax = axes[0]
candidates = list(FORMULAS.keys())
for i, name in enumerate(candidates):
    scores = np.array(fold_scores[name])
    ax.scatter([i] * len(scores), scores, color='steelblue')
    ax.hlines(scores.mean(), i - 0.15, i + 0.15, color='red', linewidth=2)
ax.set_xticks(range(len(candidates)))
ax.set_xticklabels(candidates)
ax.set_xlabel('후보 모형')
ax.set_ylabel('교차검증 포아송 이탈도')

ax = axes[1]
folds = np.arange(1, len(diff) + 1)
ax.bar(folds, diff, color='steelblue')
ax.axhline(0, color='black', linewidth=1)
ax.axhline(diff_mean, color='red', linestyle='--')
ax.set_xticks(folds)
ax.set_xlabel('폴드')
ax.set_ylabel('core-full 이탈도 차이')

plt.tight_layout()
plt.show()
```

```python
# 과제 — 프롬프트   # ai: prompt
# PROMPT 를 쓰고 → 도우미에 "위 PROMPT 대로 코드를 써 줘" → 「이 셀에 넣기」 → ▶
PROMPT = """
df, F_CORE, REG, BOOT, B, SEED는 그대로 사용해
등분산, 독립 2개의 가정으로 2개 값을 만들거야
1. 할증점수 계수의 표준오차 배수: 군집 / 포아송, 단위 배, 소수점 아래 2자리로 같은 계수에서 군집 표준오차를 포아송 표준오차로 나눈 값
2. 분산/평균이 1 위쪽인 칸수: 단위 칸-정수, 계약을 예측한 값 순으로 구간 15개로 분할
3. 함께 출력할 표 한장: 무엇으로 봤나, 진단값, 구간 길이에 매길 배수, 판정 4개 열에 가정 2개를 한행씩 기재
4. 그래프
무작위 재배치 기준 구간: 지역 라벨을 무작위로 재배치해 다시 산출한 지역별 평균들의 신용구간 범위
가로막대: 지역별 기준구간의 두 끝을 잇는 하나씩 총 22개
마커: 지역마다 실제 지역별 평균 위치에 표기
세로선: 0위치에 세로축 전체를 가로지르는 선
가로축 이름: 이탈도 잔차의 지역별 평균(전체 평균을 뺀 값)
세로축 이름:지역
"""
# --- 여기부터 AI 코드 ---
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

FCORE = FORMULAS['core']

# 모형 한 번만 정의 — offset 포함, 이후 cov_type만 바꿔 재사용
mod = smf.glm(FCORE, data=df, family=sm.families.Poisson(),
              offset=np.log(df["Exposure"]))
m_pois = mod.fit()
m_clu = mod.fit(cov_type='cluster', cov_kwds={'groups': REG})

# 1) 독립 가정 진단 — 할증점수(BonusMalus) 계수의 군집/포아송 표준오차 비
coef_name = 'BonusMalus'
se_ratio = round(m_clu.bse[coef_name] / m_pois.bse[coef_name], 2)

# 2) 등분산 가정 진단 — 예측 계약수 15분위 구간별 분산/평균비
df['_pred'] = m_pois.fittedvalues
df['_bin'] = pd.qcut(df['_pred'], q=15, duplicates='drop')
disp = df.groupby('_bin')['ClaimNb'].agg(mean='mean', var='var')
disp['ratio'] = disp['var'] / disp['mean']
n_over = int((disp['ratio'] > 1).sum())
disp_mult = round(np.sqrt(max(disp['ratio'].mean(), 1.0)), 2)
print(f"할증점수 계수의 표준오차 배수 (군집/포아송): {se_ratio}배")
print(f"분산/평균이 1 위쪽인 칸 수: {n_over}칸")

# 3) 진단 요약 표
summary = pd.DataFrame({
    '무엇으로 봤나': ['분산/평균 (구간별 과산포)', '군집/포아송 표준오차 비'],
    '진단값': [n_over, se_ratio],
    '구간 길이에 매길 배수': [disp_mult, se_ratio],
    '판정': [
        '위반' if n_over > 7 else '충족',
        '위반' if se_ratio > 1.10 else '충족',
    ],
})
print(summary)

# 4) 순열(무작위 재배치) 기준 지역별 이탈도 잔차 평균 vs 실제값
df['_resid_dev'] = m_pois.resid_deviance
overall_mean = df['_resid_dev'].mean()

codes, uniques = pd.factorize(REG)
counts = np.bincount(codes, minlength=len(uniques))
resid = df['_resid_dev'].values

rng = np.random.default_rng(SEED)
perm_means = np.empty((B, len(uniques)))
for b in range(B):
    shuffled = rng.permutation(resid)
    perm_means[b] = np.bincount(codes, weights=shuffled, minlength=len(uniques)) / counts - overall_mean

null_lo = np.percentile(perm_means, 2.5, axis=0)
null_hi = np.percentile(perm_means, 97.5, axis=0)
actual_means = np.bincount(codes, weights=resid, minlength=len(uniques)) / counts - overall_mean

plot_df = pd.DataFrame({'region': uniques, 'lo': null_lo, 'hi': null_hi, 'actual': actual_means})
plot_df = plot_df.sort_values('actual').reset_index(drop=True)

fig, ax = plt.subplots(figsize=(8, 9))
y_pos = np.arange(len(plot_df))
ax.barh(y_pos, plot_df['hi'] - plot_df['lo'], left=plot_df['lo'], color='lightgray', height=0.6, label='무작위 재배치 신용구간')
ax.scatter(plot_df['actual'], y_pos, color='red', zorder=3, label='실제 지역 평균')
ax.axvline(0, color='black', linewidth=1)
ax.set_yticks(y_pos)
ax.set_yticklabels(plot_df['region'])
ax.set_xlabel('이탈도 잔차의 지역별 평균(전체 평균을 뺀 값)')
ax.set_ylabel('지역')
ax.legend()
plt.tight_layout()
plt.show()
```
