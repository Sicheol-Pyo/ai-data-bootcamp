```python
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

DATA_DIR = Path("data")
DATA_DIR.mkdir(exist_ok=True)

def fetch_uci(url, zip_name, member):
    """UCI 정적 저장소의 zip을 내려받아 data/에 풀고, CSV 경로를 돌려줍니다."""
    csv_path = DATA_DIR / member
    if not csv_path.exists():
        zip_path = DATA_DIR / zip_name
        if not zip_path.exists():
            print(f"내려받는 중… {zip_name}")
            urllib.request.urlretrieve(url, zip_path)
        with zipfile.ZipFile(zip_path) as z:
            z.extract(member, DATA_DIR)
    return csv_path

shoppers = pd.read_csv(fetch_uci(
    "https://archive.ics.uci.edu/static/public/468/online+shoppers+purchasing+intention+dataset.zip",
    "online_shoppers.zip", "online_shoppers_intention.csv"))

NUM_COLS = ["Administrative", "Administrative_Duration", "Informational",
            "Informational_Duration", "ProductRelated", "ProductRelated_Duration",
            "BounceRates", "ExitRates", "PageValues", "SpecialDay"]
CAT_COLS = ["Month", "OperatingSystems", "Browser", "Region",
            "TrafficType", "VisitorType", "Weekend"]

# 군집화에 쓸 '행동' 피처 7개 — Revenue는 절대 포함하지 않습니다
CLUSTER_COLS = ["Administrative", "Informational", "ProductRelated",
                "ProductRelated_Duration", "BounceRates", "ExitRates", "PageValues"]

y = shoppers["Revenue"].astype(int)

print(f"세션 {len(shoppers):,}건 · 군집용 피처 {len(CLUSTER_COLS)}개")
print("\n→ 준비 완료. 이제 여러분 차례입니다.")
```

```python
#[문제 1]
#1) 군집용 7개 열을 `StandardScaler`로 표준화합니다.
scaler = StandardScaler()
X_cluster_scaler = scaler.fit_transform(shoppers[CLUSTER_COLS])
#2) `k=2~8`에서 K-Means를 학습하고 inertia와 Silhouette을 표로 만듭니다.
#   Silhouette은 계산량을 줄이기 위해 `sample_size=5000`을 사용합니다.
k_range = range(2, 9)
inertria = []
silhouette = []

for k in k_range:
    model = KMeans(n_clusters = k, n_init = 10, random_state=RANDOM_STATE).fit(X_cluster_scaler)
    inertria.append(model.inertia_)
    sil = silhouette_score(X_cluster_scaler, model.labels_, sample_size=5000, random_state=RANDOM_STATE)
    silhouette.append(sil)

result_df = pd.DataFrame({
    "k": list(k_range),
    "inertria": inertria,
    "silhouette": silhouette
})

print(result_df.round(4).to_string(index=False))

#3) 두 값을 각각 선 그래프로 그립니다.
#4) Silhouette 최댓값과 지표상 후보 범위를 기록합니다.

# [C2] 문제 1. Elbow와 Silhouette으로 `k` 후보를 좁힌다
# ⌨️ 문제 1 — k=2~8에서 inertia와 실루엣 재기
# 여기에 코드를 작성하세요 (표준화 → k별 inertia·실루엣 → 표 → 두 그래프)




```

```python
#3) 두 값을 각각 선 그래프로 그립니다.

fig, axes = plt.subplots(1, 2, figsize = (14, 4.8))

axes[0].plot(k_range, inertria, marker='o')
axes[0].set_title("Elbow")
axes[0].set_xlabel("k")
axes[0].set_ylabel("inertria")

axes[1].plot(k_range, silhouette, marker='o')
axes[1].set_title(silhouette)
axes[1].set_xlabel('k')
axes[1].set_ylabel("silhouette score")

plt.tight_layout()
plt.show()
```

```python
#4) Silhouette 최댓값과 지표상 후보 범위를 기록합니다.
best_idx = int(np.argmax(silhouette))
best_k = list(k_range)[best_idx]
print(f"\nSilhouette 최댓값: k={best_k} (score={silhouette[best_idx]:.4f})")

# 상위 몇 개를 "후보 범위"로 함께 확인
sorted_candidates = sorted(zip(k_range, silhouette), key=lambda x: x[1], reverse=True)
print("Silhouette 상위 3개 k 후보:", [(k, round(s,4)) for k, s in sorted_candidates[:3]])
```

```python
# [C3] 문제 2. k=4로 나누고, 프로파일과 2D 그림으로 확인한다
# ⌨️ 문제 2 — 군집화 → 프로파일 표 → PCA 2D 시각화
# 여기에 코드를 작성하세요 (KMeans(k=4) → 프로파일 표 → PCA 2D 산점도)

#[문제 2]
#1) k=4로 KMeans를 학습시켜 군집 라벨을 얻습니다.

model_k4 = KMeans(n_clusters=4, n_init=10, random_state=RANDOM_STATE).fit(X_cluster_scaler)
shoppers['cluster'] = model_k4.labels_
```

```python
#2) 군집별 프로파일 표를 만듭니다.
#   → 7개 피처의 평균 + 각 군집의 규모(세션 수) + 구매율
#   (구매율은 나눈 '뒤에' 보는 것이므로 규율 위반이 아닙니다)
total_sessions = shoppers.shape[0]

profile = shoppers.groupby('cluster')[CLUSTER_COLS].mean()
profile["세션 수"] = shoppers.groupby("cluster").size()
profile["구매율"] = shoppers.groupby("cluster")['Revenue'].mean() * 100
profile["세션 비율"] = profile["세션 수"] / total_sessions * 100

print(profile.round(4).to_string())
```

```python
#3) PCA로 2차원으로 줄여 군집을 색으로 구분해 그립니다.
#   설명 분산비도 함께 출력합니다.
from sklearn.decomposition import PCA

model_k4 = KMeans(n_clusters=4, n_init=10, random_state=RANDOM_STATE).fit(X_cluster_scaler)
shoppers["cluster"] = model_k4.labels_

pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_cluster_scaler)

fig, ax = plt.subplots(figsize=(9, 7))

# 군집 산점도
scatter = ax.scatter(X_pca[:, 0], X_pca[:, 1], c=shoppers["cluster"], cmap="tab10",
                     s=15, alpha=0.5)

# 원본 좌표축(0,0) 기준선 추가
ax.axhline(0, color="gray", linestyle="--", linewidth=0.8)
ax.axvline(0, color="gray", linestyle="--", linewidth=0.8)

# PC1, PC2에 대한 각 피처의 loading을 화살표로 표시 (biplot)
loadings = pca.components_.T   # shape: (원본피처개수, 2)
scale = 6   # 화살표 길이를 보기 좋게 키우는 배율 (데이터 크기에 맞게 조정)

for i, feature in enumerate(CLUSTER_COLS):
    ax.arrow(0, 0, loadings[i, 0]*scale, loadings[i, 1]*scale,
             color="red", alpha=0.8, head_width=0.15, linewidth=1.2)
    ax.text(loadings[i, 0]*scale*1.15, loadings[i, 1]*scale*1.15, feature,
            color="darkred", fontsize=9, ha="center")

ax.set_xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)")
ax.set_ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)")
ax.set_title(f"군집 시각화 + 피처 방향(biplot) (k=4, 설명분산비 {pca.explained_variance_ratio_.sum()*100:.1f}%)")
plt.colorbar(scatter, ax=ax, label="군집 번호")
plt.tight_layout()
plt.show()
```

```python
# [C5] 문제 4. 군집 중심거리 피처를 예측 모델에 재투입한다
# ⌨️ 문제 4 — 없음 / Pipeline 안 중심거리 / 전체-fit 중심거리 3방향 비교
# 여기에 코드를 작성하세요 (3방향 CV → 차이 계산 → 채택 결정)

#[문제 4]
#1) 세 방식을 같은 층화 5겹 CV로 평가합니다. 모델은 순서 5의 RandomForest(`min_samples_leaf=5`)입니다.
#   ① 군집 파생 피처 없음
#   ② `Pipeline` 안에서 표준화 → KMeans(4) 중심거리 생성(정상)
#   ③ 전체 데이터로 미리 fit한 표준화·KMeans 중심거리 사용(누수)
#2) F1과 AP 차이를 계산합니다.
#3) 군집 파생 피처 채택 여부를 정하고 근거를 기록합니다.
#4) 효과가 관찰되지 않았다면 데이터와 모델 구조로 설명합니다.

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline, FeatureUnion
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestClassifier

X_full = shoppers[NUM_COLS + CAT_COLS]  # 실제 사용할 전체 피처 구성에 맞게 조정
y = shoppers["Revenue"].astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X_full, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
scoring = ["f1", "average_precision"]

def rf():
    return RandomForestClassifier(n_estimators=300, min_samples_leaf=5, random_state=RANDOM_STATE)

# ① 군집 파생 피처 없음
plain = Pipeline([("scale", StandardScaler()), ("clf", rf())])
res_plain = cross_validate(plain, X_train[CLUSTER_COLS], y_train, cv=cv, scoring=scoring)

# ② Pipeline 안에서 표준화 → KMeans(4) 중심거리 생성 (정상, 누수 없음)
feat_inside = FeatureUnion([
    ("original", "passthrough"),
    ("cluster_distance", KMeans(n_clusters=4, n_init=10, random_state=RANDOM_STATE)),
])
inside = Pipeline([("scale", StandardScaler()), ("feat", feat_inside), ("clf", rf())])
res_inside = cross_validate(inside, X_train[CLUSTER_COLS], y_train, cv=cv, scoring=scoring)

# ③ 전체 데이터로 미리 fit한 표준화·KMeans 중심거리 사용 (누수)
scaler_all = StandardScaler().fit(X_train[CLUSTER_COLS])
X_scaled_all = scaler_all.transform(X_train[CLUSTER_COLS])
km_all = KMeans(n_clusters=4, n_init=10, random_state=RANDOM_STATE).fit(X_scaled_all)
X_leak = np.column_stack([X_scaled_all, km_all.transform(X_scaled_all)])
res_leak = cross_validate(rf(), X_leak, y_train, cv=cv, scoring=scoring)

# 2) 결과 정리 및 F1·AP 차이 계산
summary = pd.DataFrame({
    "방식": ["① 군집 없음", "② Pipeline 안(정상)", "③ 전체-fit(누수)"],
    "F1": [res_plain["test_f1"].mean(), res_inside["test_f1"].mean(), res_leak["test_f1"].mean()],
    "AP": [res_plain["test_average_precision"].mean(), res_inside["test_average_precision"].mean(), res_leak["test_average_precision"].mean()],
})
summary["F1 변화(대비①)"] = summary["F1"] - summary.loc[0, "F1"]
summary["AP 변화(대비①)"] = summary["AP"] - summary.loc[0, "AP"]

print(summary.round(4).to_string(index=False))

print(f"\n정상 절차의 피처화 효과 (②-①): F1 {summary.loc[1,'F1']-summary.loc[0,'F1']:+.4f}, "
      f"AP {summary.loc[1,'AP']-summary.loc[0,'AP']:+.4f}")
print(f"누수로 인한 부풀림 (③-②): F1 {summary.loc[2,'F1']-summary.loc[1,'F1']:+.4f}, "
      f"AP {summary.loc[2,'AP']-summary.loc[1,'AP']:+.4f}")
```

```python
f1_effect = summary.loc[1, "F1"] - summary.loc[0, "F1"]
ap_effect = summary.loc[1, "AP"] - summary.loc[0, "AP"]

print("\n=== 채택 여부 판단 ===")
if f1_effect > 0.01 and ap_effect > 0.01:
    print("채택: 군집 파생 피처가 F1·AP 모두 뚜렷하게 개선시켰다.")
elif f1_effect <= 0 and ap_effect <= 0:
    print("기각: 군집 파생 피처가 성능을 개선시키지 못했다.")
    print("설명: RandomForest는 원래 '거리 기반 정보' 없이도, 원본 피처들의 조합으로 "
          "비슷한 비선형 경계를 스스로 학습할 수 있는 모델이라, 군집 중심거리라는 "
          "요약 정보가 추가적인 새로운 신호를 주지 못했을 가능성이 높다.")
else:
    print("판단 보류: 두 지표가 다른 방향을 가리켜, 추가 검증(다른 k, 다른 모델)이 필요하다.")
```
