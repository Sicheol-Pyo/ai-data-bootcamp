```python
# ─────────────────────────────────────────────
# [C1] ⚙️ 데이터 준비
# 최초 1회 다운로드 → data/ 폴더에 저장 (이후 오프라인)
# ─────────────────────────────────────────────
import urllib.request, zipfile
from pathlib import Path

import numpy as np
import pandas as pd

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

DATA_DIR = Path("data")
DATA_DIR.mkdir(exist_ok=True)

def fetch_uci(url, zip_name, member):
    """UCI 정적 저장소의 zip을 내려받아 data/에 풀고, CSV 경로를 돌려줍니다."""
    csv_path = DATA_DIR / member
    if not csv_path.exists():
        zip_path = DATA_DIR / zip_name
        if not zip_path.exists():
            print(f"내려받는 중… {zip_name}")
            urllib.request.urlretrieve(url, zip_path)
        with zipfile.ZipFile(zip_path) as z:
            z.extract(member, DATA_DIR)
    return csv_path

shoppers = pd.read_csv(fetch_uci(
    "https://archive.ics.uci.edu/static/public/468/online+shoppers+purchasing+intention+dataset.zip",
    "online_shoppers.zip", "online_shoppers_intention.csv"))

NUM_COLS = ["Administrative", "Administrative_Duration", "Informational",
            "Informational_Duration", "ProductRelated", "ProductRelated_Duration",
            "BounceRates", "ExitRates", "PageValues", "SpecialDay"]
CAT_COLS = ["Month", "OperatingSystems", "Browser", "Region",
            "TrafficType", "VisitorType", "Weekend"]

y = shoppers["Revenue"].astype(int)

print(f"수치형 {len(NUM_COLS)}개 · 범주형 {len(CAT_COLS)}개 · 타깃 1개")
print("범주별 값 개수:", {c: shoppers[c].nunique() for c in CAT_COLS})
print("\n→ 준비 완료. 이제 여러분 차례입니다.")
```

```python
#[문제 1]
#1) ColumnTransformer를 만듭니다.
#   - 수치형: "passthrough"
#   - 범주형: OneHotEncoder(handle_unknown="ignore", min_frequency=20)
#2) 그것과 지난 순서의 모델을 Pipeline으로 묶습니다.
#   RandomForestClassifier(n_estimators=300, min_samples_leaf=20)
#3) 5겹 CV로 F1과 AP(`average_precision`)를 측정하고, 수치형 10개만 썼을 때와 비교합니다.
#4) 변환 후 피처가 몇 개가 됐는지 출력합니다.

# [C2] 문제 1. 범주형 7개 열을 `Pipeline`으로 푼다
# ⌨️ 문제 1 — ColumnTransformer + Pipeline으로 범주형 개방
# 여기에 코드를 작성하세요 (ColumnTransformer → Pipeline → CV 비교 → 피처 수)
X = shoppers[NUM_COLS + CAT_COLS]
y = shoppers["Revenue"].astype(int)

X_train, X_test, y_train, y_test =train_test_split(
    X, y, test_size = 0.2, stratify= y, random_state = 42)

preprocess = ColumnTransformer([
    ("num", "passthrough", NUM_COLS),
    ("cat", OneHotEncoder(handle_unknown="ignore", min_frequency=20), CAT_COLS)
])

pipe = Pipeline([
    ("pre", preprocess),
    ("clf", RandomForestClassifier(n_estimators=300, min_samples_leaf=20, random_state=42))
])

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scoring = ["f1", "average_precision"]

results = cross_validate(pipe, X_train, y_train, cv = skf, scoring=scoring)

for metric in scoring :
    print(f"{metric}: {results[f'test_{metric}'].mean():.4f}")

preprocess.fit(X_train)
n_features_after = preprocess.transform(X_train).shape[1]
print(f"\n변환 후 피처 개수: {n_features_after}개")
```

```python
#[문제 2]
#1) 아래 두 가설을 코드로 만들기 전에, 각각 "왜 효과가 있을 것인가"를 한 줄로 적습니다.
#   가설 A: has_page_value = (PageValues > 0)  ← 0 여부가 추가 표현으로 유용할 수 있다
#   가설 B: total_pages = Administrative + Informational + ProductRelated  ← 총 탐색량
#2) 문제 1의 ② 구성을 기준으로, A만 / B만 각각 추가해 5겹 CV로 측정합니다.
#3) 각각의 ΔF1과 ΔAP를 계산해 채택·기각을 결정합니다.

# [C3] 문제 2. 파생변수 — 가설을 먼저 적고, 그다음 검증한다
# ⌨️ 문제 2 — 가설 A·B를 각각 따로 검증

# 여기에 코드를 작성하세요 (파생 열 추가 → A만 / B만 → Δ 계산 → 채택 결정)

shoppers["has_page_value"] = (shoppers["PageValues"] > 0).astype(int)
shoppers["total_pages"] = shoppers["Administrative"] + shoppers["Informational"] + shoppers["ProductRelated"]

X = shoppers[NUM_COLS + CAT_COLS + ["has_page_value", "total_pages"]]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

def cv_of(num_cols, cat_cols):
    pre = ColumnTransformer([
        ("num", "passthrough", num_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore", min_frequency=20), cat_cols)
    ])
    pipe = Pipeline([
        ("pre", pre),
        ("clf", RandomForestClassifier(n_estimators=300, min_samples_leaf=20, random_state=42))
    ])
    results = cross_validate(pipe, X_train[num_cols + cat_cols], y_train, cv=skf, scoring=scoring)
    return {m: results[f"test_{m}"].mean() for m in scoring}


base = cv_of(NUM_COLS, CAT_COLS)
print(f"② 기준: F1={base['f1']:.4f}, AP={base['average_precision']:.4f}")

# 가설 A만 추가
score_a = cv_of(NUM_COLS + ["has_page_value"], CAT_COLS)
delta_a_f1 = score_a["f1"] - base["f1"]
delta_a_ap = score_a["average_precision"] - base["average_precision"]
print(f"가설 A 추가: F1={score_a['f1']:.4f}(Δ{delta_a_f1:+.4f}), AP={score_a['average_precision']:.4f}(Δ{delta_a_ap:+.4f})")

# 가설 B만 추가
score_b = cv_of(NUM_COLS + ["total_pages"], CAT_COLS)
delta_b_f1 = score_b["f1"] - base["f1"]
delta_b_ap = score_b["average_precision"] - base["average_precision"]
print(f"가설 B 추가: F1={score_b['f1']:.4f}(Δ{delta_b_f1:+.4f}), AP={score_b['average_precision']:.4f}(Δ{delta_b_ap:+.4f})")

# 3) 채택/기각 결정
print("\n=== 채택/기각 결정 ===")
for name, d_f1, d_ap in [("가설 A(has_page_value)", delta_a_f1, delta_a_ap),
                          ("가설 B(total_pages)", delta_b_f1, delta_b_ap)]:
    decision = "채택" if (d_f1 > 0 and d_ap > 0) else "기각"
    print(f"{name}: {decision} (ΔF1={d_f1:+.4f}, ΔAP={d_ap:+.4f})")
```

```python
# [C4] 문제 3. 누수를 직접 만들어, 부풀림을 잰다
# ⌨️ 문제 3 — 같은 피처를 누수 버전과 정상 버전으로 각각 재기
from sklearn.preprocessing import TargetEncoder

# 여기에 코드를 작성하세요 (조합 키 → 누수 버전 CV → Pipeline 안 버전 CV → 차이)

#[문제 3]
#1) 조합 키를 만듭니다 — combo = Month + "_" + TrafficType + "_" + Region
#   (몇 개 조합이 나오는지, 10건 미만인 조합이 몇 개인지 확인합니다)
#2) 누수 버전: 전체 데이터로 조합별 Revenue 평균을 구해 열로 추가 → CV
#3) 교차 적합 버전: 같은 일을 `TargetEncoder`로 `Pipeline` 안에서 수행 → CV
#4) 두 점수의 차이를 계산하고, 교차 적합 버전이 기준(문제 1의 ②)보다 나아졌는지 확인합니다.

# 1) 조합 키 만들기
shoppers["combo"] = (shoppers["Month"].astype(str) + "_" +
                     shoppers["TrafficType"].astype(str) + "_" +
                     shoppers["Region"].astype(str))

n_combos = shoppers["combo"].nunique()
combo_counts = shoppers["combo"].value_counts()
n_small = (combo_counts < 10).sum()
print(f"전체 조합 개수: {n_combos}개")
print(f"10건 미만인 조합 개수: {n_small}개")

X_full = shoppers[NUM_COLS + CAT_COLS + ["combo"]]
X_train, X_test, y_train, y_test = train_test_split(X_full, y, test_size=0.2, stratify=y, random_state=42)

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 2) 누수 버전: 전체 데이터로 조합별 Revenue 평균을 미리 계산해서 열로 추가
combo_target_mean_leak = shoppers.groupby("combo")["Revenue"].mean()   # 전체(train+test)로 계산 → 누수!
X_train_leak = X_train.copy()
X_train_leak["combo_encoded_leak"] = X_train_leak["combo"].map(combo_target_mean_leak)

pre_leak = ColumnTransformer([
    ("num", "passthrough", NUM_COLS + ["combo_encoded_leak"]),
    ("cat", OneHotEncoder(handle_unknown="ignore", min_frequency=20), CAT_COLS)
])
pipe_leak = Pipeline([
    ("pre", pre_leak),
    ("clf", RandomForestClassifier(n_estimators=300, min_samples_leaf=20, random_state=42))
])
score_leak = cross_val_score(
    pipe_leak, X_train_leak[NUM_COLS + CAT_COLS + ["combo_encoded_leak"]], y_train, cv=skf, scoring="f1"
).mean()
print(f"누수 버전 F1: {score_leak:.4f}")

# 3) 교차 적합 버전: TargetEncoder를 Pipeline 안에 포함 (자동으로 누수 방지 처리됨)
pre_correct = ColumnTransformer([
    ("num", "passthrough", NUM_COLS),
    ("cat", OneHotEncoder(handle_unknown="ignore", min_frequency=20), CAT_COLS),
    ("combo", TargetEncoder(random_state=42), ["combo"])
])
pipe_correct = Pipeline([
    ("pre", pre_correct),
    ("clf", RandomForestClassifier(n_estimators=300, min_samples_leaf=20, random_state=42))
])
score_correct = cross_val_score(
    pipe_correct, X_train[NUM_COLS + CAT_COLS + ["combo"]], y_train, cv=skf, scoring="f1"
).mean()
print(f"교차 적합(TargetEncoder in Pipeline) 버전 F1: {score_correct:.4f}")
print(f"누수로 인한 부풀림: {score_leak - score_correct:+.4f}")

# 4) 기준(② NUM+CAT)과 비교
pre_base = ColumnTransformer([
    ("num", "passthrough", NUM_COLS),
    ("cat", OneHotEncoder(handle_unknown="ignore", min_frequency=20), CAT_COLS)
])
pipe_base = Pipeline([
    ("pre", pre_base),
    ("clf", RandomForestClassifier(n_estimators=300, min_samples_leaf=20, random_state=42))
])
score_base = cross_val_score(pipe_base, X_train[NUM_COLS + CAT_COLS], y_train, cv=skf, scoring="f1").mean()

print(f"기준(② NUM+CAT) F1: {score_base:.4f}")
print(f"정상 버전(combo 추가) F1: {score_correct:.4f}")
print(f"개선 여부: {'개선됨' if score_correct > score_base else '개선 안됨'} ({score_correct - score_base:+.4f})")
```

```python
# [C5] 문제 4. `Pipeline` 통째로 `GridSearchCV`
# ⌨️ 문제 4 — 전처리 옵션과 모델 옵션을 한 격자에서
# 여기에 코드를 작성하세요 (격자 정의 → GridSearchCV → 결과표 → best)

#[문제 4]
#1) 문제 2에서 채택한 구성(수치 10개 + has_page_value + 범주 7개)으로 Pipeline을 만듭니다.
#2) GridSearchCV로 다음 두 손잡이를 함께 탐색합니다. `scoring="average_precision"`
#   - pre__cat__min_frequency: [10, 50]      ← 전처리 옵션
#   - clf__min_samples_leaf:   [5, 20, 50]   ← 모델 옵션
#3) 조합별 내부 CV 점수를 표로 출력하고, `best_params_`와 `best_score_`를 확인합니다.
#4) 선택된 모델의 F1도 같은 5겹 분할에서 측정해 기록합니다. 이 값은 모델 선택에 사용한 데이터의 내부 추정치이므로 최종 일반화 성능으로 보고하지 않습니다.

# 0) 문제 2에서 채택한 구성으로 데이터 준비
FEAT_NUM = NUM_COLS + ["has_page_value"]

X_full = shoppers[FEAT_NUM + CAT_COLS]
y = shoppers["Revenue"].astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X_full, y, test_size=0.2, stratify=y, random_state=42
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 1) Pipeline 만들기
pre = ColumnTransformer([
    ("num", "passthrough", FEAT_NUM),
    ("cat", OneHotEncoder(handle_unknown="ignore"), CAT_COLS)   # min_frequency는 grid에서 탐색할 것이라 여기선 기본만
])
pipe = Pipeline([
    ("pre", pre),
    ("clf", RandomForestClassifier(n_estimators=300, random_state=42))
])

# 2) GridSearchCV — 전처리 옵션 + 모델 옵션 동시 탐색
param_grid = {
    "pre__cat__min_frequency": [10, 50],
    "clf__min_samples_leaf": [5, 20, 50]
}

gs = GridSearchCV(pipe, param_grid, cv=cv, scoring="average_precision", n_jobs=-1)
gs.fit(X_train, y_train)

# 3) 결과표 출력
results_df = pd.DataFrame(gs.cv_results_)[
    ["param_pre__cat__min_frequency", "param_clf__min_samples_leaf", "mean_test_score", "std_test_score"]
]
results_df = results_df.sort_values("mean_test_score", ascending=False)
print("=== 조합별 CV 점수 (AP 기준) ===")
print(results_df.round(4).to_string(index=False))

print(f"\nbest_params_: {gs.best_params_}")
print(f"best_score_ (AP): {gs.best_score_:.4f}")

# 4) 선택된 모델의 F1도 같은 5겹 분할에서 측정 (내부 추정치임을 명시)
f1_scores = cross_val_score(gs.best_estimator_, X_train, y_train, cv=cv, scoring="f1")
print(f"\n선택된 모델의 F1 (같은 5겹 CV, 내부 추정치): {f1_scores.mean():.4f}")
print("※ 이 값은 모델 선택에 이미 사용된 데이터에서 나온 추정치이므로,")
print("  최종 일반화 성능으로 보고하지 않고, 반드시 X_test로 별도 검증해야 함")
```
