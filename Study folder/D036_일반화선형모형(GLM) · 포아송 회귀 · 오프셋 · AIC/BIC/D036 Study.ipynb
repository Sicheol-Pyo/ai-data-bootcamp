```python
# 미션 1 · 준비 — 읽고 실행만 합니다. 이 셀이 미션 1 에 쓸 변수를 직접 만듭니다
from sklearn.datasets import fetch_openml

# freMTPL2freq — 프랑스 자동차보험 계약별 사고 빈도 (678,013 x 12). OpenML 에서 바로 내려받습니다
df = fetch_openml(data_id=41214, as_frame=True).frame
df["IDpol"] = df["IDpol"].astype(int)
for c in ["Area", "VehBrand", "VehGas", "Region"]:
    df[c] = df[c].astype(str).str.strip("'").astype("category")

# 첫날 정해 닷새 내내 고정하는 전처리 두 규칙
df["ClaimNb"] = df["ClaimNb"].astype(int).clip(upper=4)   # 사고 5건 이상은 극소수라 상한 4 로 상한 처리
df["Exposure"] = df["Exposure"].clip(upper=1.0)           # 노출은 1년 상한

off = np.log(df["Exposure"])                  # 쓸 변수 1 — 노출의 로그 (offset 에 넘길 값, 노출 1년이면 0)
CLAIMS_ALL = int(df["ClaimNb"].sum())         # 쓸 변수 2 — 총 사고 건수 (단위 건)
EXP_ALL = float(df["Exposure"].sum())         # 쓸 변수 3 — 총 노출 (단위 년)

bm_edges = [50, 55, 60, 70, 80, 90, 100, 110, 125, 150, 230]        # 쓸 변수 4 — 할증점수 구간 경계 (점)
_bm_cut = pd.cut(df["BonusMalus"], bm_edges, right=False)
_bm_sum = df.groupby(_bm_cut, observed=True)[["ClaimNb", "Exposure"]].sum()
bm_x = [(a + b) / 2 for a, b in zip(bm_edges[:-1], bm_edges[1:])]   # 쓸 변수 5 — 구간 중앙값 (그림의 가로축)
bm_rate = (_bm_sum["ClaimNb"] / _bm_sum["Exposure"]).to_numpy()     # 쓸 변수 6 — 구간별 관측 연간 사고율 (건/년)

print("행 x 열 :", df.shape, "| 결측치 총합 :", int(df.isna().sum().sum()))
print(f"총 사고 {CLAIMS_ALL:,} 건 | 총 노출 {EXP_ALL:,.1f} 년 |"
      f" log(Exposure) {off.min():.3f} ~ {off.max():.3f}  (노출 1년이면 0)")
print(f"할증점수 구간 {len(bm_x)}개 | 구간별 관측 사고율 {bm_rate.min():.4f} ~ {bm_rate.max():.4f} 건/년")
print("쓸 수 있는 이름 : df · off (노출의 로그) · CLAIMS_ALL (건) · EXP_ALL (년) ·"
      " bm_x (점) · bm_rate (건/년)")
```

```python
# 미션 1 — 프롬프트   # ai: prompt
# PROMPT 를 쓰고 → 도우미에 "위 PROMPT 대로 코드를 써 줘" → 「이 셀에 넣기」 → ▶
PROMPT = """
준비 셀에서 만들어둔 항목을 따로 추가로 만들지 말고 해당 변수를 그대로 사용해줘
df, off, bm_x, bm_rate
아래 내가 기재한 항목을 순차적으로 만들거야
1. 연간 사고율: 단위 건/년, 소수점 여섯자리까지 (절편만 둔 포아송 GLM의 절편을 지수로 취해 복원한 값)
2. null 모형의 로그우도: 단위 없음, 소수점 아래 두자리, 적합이 최대화한 값
3. 로그 스케일 그대로의 절편 (추가 행)
4. 그래프
가로축: 할증점수(구간 중앙값)
세로축: 연간 사고율(건/년, 로그눈금)
꺽은선: bm_x와 bm_rate를 이은 선 하나, 범례 문구는 '구간별 관측 사고율'
가로선: 연간 사고율 값에 맞춰 가로축 전체를 가로지르는 선 하나 'null 모형이 매기는 단일 요율'
세로축 눈금: 로그 눈금

"""
# --- 여기부터 AI 코드 ---
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

# 1. 절편만 둔 포아송 GLM (오프셋 사용)
X_null = np.ones((len(df), 1))
model_null = sm.GLM(df['ClaimNb'], X_null,
                     family=sm.families.Poisson(),
                     offset=off)
result_null = model_null.fit()

intercept = result_null.params.iloc[0]
annual_rate = np.exp(intercept)
print(f"연간 사고율: {annual_rate:.6f} 건/년")

# 2. null 모형의 로그우도
loglik_null = result_null.llf
print(f"null 모형 로그우도: {loglik_null:.2f}")

# 3. 로그 스케일 그대로의 절편
print(f"로그 스케일 절편: {intercept:.6f}")

# 4. 그래프
plt.plot(bm_x, bm_rate, marker='o', label='구간별 관측 사고율')
plt.axhline(annual_rate, color='red', linestyle='--', label='null 모형이 매기는 단일 요율')
plt.xlabel('할증점수(구간 중앙값)')
plt.ylabel('연간 사고율(건/년, 로그눈금)')
plt.yscale('log')
plt.legend()
plt.show()
```

```python
# 미션 3 — 프롬프트   # ai: prompt
# PROMPT 를 쓰고 → 도우미에 "위 PROMPT 대로 코드를 써 줘" → 「이 셀에 넣기」 → ▶
PROMPT = """
df, FORMULAS, F_CORE, F_FULL, off는 그대로 사용해줘
후보 3개의 비교와 요율표 선정을 진행하려고 해
1. 가장 낮은 AIC: 단위 점, 소수점 아래 2자리, 후보 3개의 AIC 중 가장 낮은 값, 그리고 어떤 AIC(모형)인지도 기재해줘
2. 채점표 null 행의 로그우도: 단위 없음, 소수점 아래 2자리, 적합이 최대화한 값(음수)
3. 채점표 한장: 한 행이 후보 하나이고, 열은 이름, 계수 개수, 로그우도, 이탈도, AIC, BIC, 적합게 쓰인 관측 수
4. 그래프(가로축은 후보 이름, 세로축은 AIC인 세로 막대 그래프)
막대: 후보 하나에 막대 하나, 총 3개(가로축 눈금 null, core, full)
강조: 가장 낮은 막대만 색 다르게
세로축 눈금: 세 막대의 차이가 보이도록 잘라서
가로축 이름: 후보 모형
세로축 이름: AIC (점, 낮을수록 좋다)
"""
# --- 여기부터 AI 코드 ---
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import statsmodels.formula.api as smf

rows = []
for name, formula in FORMULAS.items():
    res = smf.glm(formula=formula, data=df, family=sm.families.Poisson(), offset=off).fit()
    rows.append({
        "이름": name,
        "계수 개수": int(res.df_model) + 1,
        "로그우도": res.llf,
        "이탈도": res.deviance,
        "AIC": res.aic,
        "BIC": res.bic,
        "관측 수": int(res.nobs),
    })

score_table = pd.DataFrame(rows).set_index("이름")

best_name = score_table["AIC"].idxmin()
best_aic = score_table["AIC"].min()
print(f"가장 낮은 AIC = {best_aic:.2f} 점 ({best_name})")

null_loglik = score_table.loc["null", "로그우도"]
print(f"null 로그우도 = {null_loglik:.2f}")

print(score_table.round(2).to_string())

# 그래프
colors = ['tab:orange' if name == best_name else 'tab:blue' for name in score_table.index]
plt.figure(figsize=(5, 4))
plt.bar(score_table.index, score_table["AIC"], color=colors)
plt.ylim(score_table["AIC"].min() - 50, score_table["AIC"].max() + 50)
plt.xlabel("후보 모형")
plt.ylabel("AIC (점, 낮을수록 좋다)")
plt.show()
```
