```python
#transcript[value] key 'offer id' -> 'offer_id'로 수정

def normalize_value(v):
    if 'offer id' in v:
        v = dict(v)  # 원본 dict 훼손 방지
        v['offer_id'] = v.pop('offer id')
    return v

transcript['value'] = transcript['value'].apply(normalize_value)
```

```python
# 1. 소비 부담 대비 낮은 Reward(보상)
# --- 준비 ---
portfolio_info = portfolio.set_index('id').to_dict('index')
bogo_discount_ids = set(portfolio.loc[portfolio['offer_type'].isin(['bogo','discount']), 'id'])

def get_offer_id(v):
    return v.get('offer_id')

transcript['offer_id'] = transcript['value'].apply(get_offer_id)

# --- person x offer_id 단위로 발송 인스턴스별 수신/완료 매칭 ---
records = []
for person, g in transcript.groupby('person', sort=False):
    g_bd = g[g['offer_id'].isin(bogo_discount_ids)]
    if g_bd.empty:
        continue
    received = g_bd[g_bd['event']=='offer received'].sort_values('time')
    viewed = g_bd[g_bd['event']=='offer viewed'].sort_values('time')
    completed = g_bd[g_bd['event']=='offer completed'].sort_values('time')

    for oid, r_group in received.groupby('offer_id'):
        duration_hours = portfolio_info[oid]['duration'] * 24
        v_list = viewed.loc[viewed['offer_id']==oid, 'time'].tolist()
        c_list = completed.loc[completed['offer_id']==oid, 'time'].tolist()
        used_v = [False]*len(v_list)
        used_c = [False]*len(c_list)
        for rt in r_group['time'].tolist():
            window_end = rt + duration_hours
            is_viewed = False
            view_time = None
            for i, vt in enumerate(v_list):
                if not used_v[i] and rt <= vt <= window_end:
                    used_v[i] = True
                    is_viewed = True
                    view_time = vt
                    break
            is_completed_after_view = False
            if is_viewed:
                for i, ct in enumerate(c_list):
                    if not used_c[i] and view_time <= ct <= window_end:
                        used_c[i] = True
                        is_completed_after_view = True
                        break
            records.append({
                'person': person, 'offer_id': oid,
                'viewed': is_viewed,
                'viewed_before_completed': is_completed_after_view,
                'reward': portfolio_info[oid]['reward'],
                'difficulty': portfolio_info[oid]['difficulty'],
                'offer_type': portfolio_info[oid]['offer_type'],
                'duration': portfolio_info[oid]['duration'],
            })

instances = pd.DataFrame(records)

# --- 로지스틱 회귀 (수신한 인스턴스만 대상) ---
model_df = instances[instances['viewed']].copy()
model_df['viewed_before_completed'] = model_df['viewed_before_completed'].astype(int)  # bool -> 0/1, statsmodels 포뮬러용

model = smf.logit('viewed_before_completed ~ reward + difficulty', data=model_df).fit()
print(model.summary())
```

```python
# 3. Mobile, Social 대비 Web, Email 사용률/도달률이 낮음

def get_offer_id(v):
    return v.get('offer_id')

transcript['offer_id'] = transcript['value'].apply(get_offer_id)
portfolio_info = portfolio.set_index('id').to_dict('index')
all_offer_ids = set(portfolio['id'])  # 이 가설은 완료 여부와 무관하니 informational 포함 10개 전부 사용

# --- person x offer_id 단위로 발송->수신 매칭 (완료는 필요 없음) ---
reach_records = []
for person, g in transcript.groupby('person', sort=False):
    g_all = g[g['offer_id'].isin(all_offer_ids)]
    if g_all.empty:
        continue
    received = g_all[g_all['event']=='offer received'].sort_values('time')
    viewed = g_all[g_all['event']=='offer viewed'].sort_values('time')

    for oid, r_group in received.groupby('offer_id'):
        duration_hours = portfolio_info[oid]['duration'] * 24
        v_list = viewed.loc[viewed['offer_id']==oid, 'time'].tolist()
        used_v = [False]*len(v_list)
        for rt in r_group['time'].tolist():
            window_end = rt + duration_hours
            is_viewed = False
            for i, vt in enumerate(v_list):
                if not used_v[i] and rt <= vt <= window_end:
                    used_v[i] = True
                    is_viewed = True
                    break
            reach_records.append({'person': person, 'offer_id': oid, 'viewed': is_viewed})

reach = pd.DataFrame(reach_records)

# --- 오퍼 단위로 채널 구성 + 수신율 집계 ---
def channel_flags(offer_id):
    ch = portfolio_info[offer_id]['channels']
    return pd.Series({
        'has_mobile': 'mobile' in ch,
        'has_social': 'social' in ch,
        'has_web': 'web' in ch,
        'has_email': 'email' in ch,
    })

offer_summary = (
    reach.groupby('offer_id')['viewed']
    .agg(발송수='size', 수신수='sum')
    .assign(수신율=lambda d: d['수신수'] / d['발송수'])
    .reset_index()
)
flags = offer_summary['offer_id'].apply(channel_flags)
offer_summary = pd.concat([offer_summary, flags], axis=1)
offer_summary['offer_type'] = offer_summary['offer_id'].map(lambda o: portfolio_info[o]['offer_type'])
offer_summary['channels'] = offer_summary['offer_id'].map(lambda o: portfolio_info[o]['channels'])

offer_summary = offer_summary.sort_values('수신율', ascending=False)
print(offer_summary[['offer_id', 'offer_type', 'channels', '발송수', '수신수', '수신율']])

# --- 채널 포함 여부별 평균 수신율 비교 ---
for ch in ['has_web', 'has_email', 'has_mobile', 'has_social']:
    포함 = offer_summary.loc[offer_summary[ch], '수신율'].mean()
    미포함 = offer_summary.loc[~offer_summary[ch], '수신율'].mean()
    print(f"{ch}: 포함 오퍼 평균수신율={포함:.4f} (n={offer_summary[ch].sum()}개) / 미포함={미포함:.4f} (n={(~offer_summary[ch]).sum()}개)")
```

```python
# 5. Duration이 짧은 경우 offer 인지 전 마감될 수 있음
# --- person x offer_id 단위로 발송~수신 매칭, 시각 정보까지 포함해서 재구성 ---
# (all_offer_ids, portfolio_info는 4번 가설 코드에서 만든 걸 그대로 재사용 — informational 포함 10개 전부)
from scipy.stats import pointbiserialr

records = []
for person, g in transcript.groupby('person', sort=False):
    g_all = g[g['offer_id'].isin(all_offer_ids)]
    if g_all.empty:
        continue
    received = g_all[g_all['event']=='offer received'].sort_values('time')
    viewed = g_all[g_all['event']=='offer viewed'].sort_values('time')

    for oid, r_group in received.groupby('offer_id'):
        duration_hours = portfolio_info[oid]['duration'] * 24
        v_list = viewed.loc[viewed['offer_id']==oid, 'time'].tolist()
        used_v = [False] * len(v_list)
        for rt in r_group['time'].tolist():
            window_end = rt + duration_hours
            is_viewed = False
            view_time = np.nan
            for i, vt in enumerate(v_list):
                if not used_v[i] and rt <= vt <= window_end:
                    used_v[i] = True
                    is_viewed = True
                    view_time = vt
                    break
            records.append({
                'person': person,
                'offer_id': oid,
                'received_time': rt,
                'viewed_time': view_time,
                'viewed': is_viewed,
                'duration': portfolio_info[oid]['duration'],
            })

duration_df = pd.DataFrame(records)
duration_df['viewed_int'] = duration_df['viewed'].astype(int)

# --- 상관분석: duration ↔ viewed(0/1) ---
corr, p = pointbiserialr(duration_df['duration'], duration_df['viewed_int'])
print(f"duration-viewed 상관계수: {corr:.4f}, p-value: {p:.4g}, N={len(duration_df)}")

# --- 로지스틱 회귀: viewed ~ duration ---
model_5 = smf.logit('viewed_int ~ duration', data=duration_df).fit()
print(model_5.summary())
```

```python
# 6. Offer 노출 횟수에 따라 반응 둔화
# --- person x offer_id 단위 인스턴스 테이블 (전체 10개 오퍼, viewed + bogo/discount는 완료까지) ---
# all_offer_ids, bogo_discount_ids, portfolio_info는 앞서 만든 걸 재사용
records = []
for person, g in transcript.groupby('person', sort=False):
    g_all = g[g['offer_id'].isin(all_offer_ids)]
    if g_all.empty:
        continue
    received = g_all[g_all['event']=='offer received'].sort_values('time')
    viewed = g_all[g_all['event']=='offer viewed'].sort_values('time')
    completed = g_all[g_all['event']=='offer completed'].sort_values('time')

    for oid, r_group in received.groupby('offer_id'):
        duration_hours = portfolio_info[oid]['duration'] * 24
        is_bd = oid in bogo_discount_ids
        v_list = viewed.loc[viewed['offer_id']==oid, 'time'].tolist()
        c_list = completed.loc[completed['offer_id']==oid, 'time'].tolist() if is_bd else []
        used_v = [False] * len(v_list)
        used_c = [False] * len(c_list)
        for rt in r_group['time'].tolist():
            window_end = rt + duration_hours
            is_viewed = False
            view_time = None
            for i, vt in enumerate(v_list):
                if not used_v[i] and rt <= vt <= window_end:
                    used_v[i] = True
                    is_viewed = True
                    view_time = vt
                    break
            is_completed_after_view = False
            if is_viewed and is_bd:
                for i, ct in enumerate(c_list):
                    if not used_c[i] and view_time <= ct <= window_end:
                        used_c[i] = True
                        is_completed_after_view = True
                        break
            records.append({
                'person': person,
                'offer_id': oid,
                'received_time': rt,
                'viewed': is_viewed,
                'viewed_before_completed': is_completed_after_view if is_bd else np.nan,
                'is_bogo_discount': is_bd,
            })

fatigue_df = pd.DataFrame(records)

# --- prior_offer_count: 이번 수신 이전까지 그 사람이 받은 오퍼 수 (시간순 0부터) ---
fatigue_df = fatigue_df.sort_values(['person', 'received_time'])
fatigue_df['prior_offer_count'] = fatigue_df.groupby('person').cumcount()
fatigue_df['viewed_int'] = fatigue_df['viewed'].astype(int)

# --- 상관분석 1: prior_offer_count ↔ viewed ---
corr_viewed, p_viewed = pointbiserialr(fatigue_df['prior_offer_count'], fatigue_df['viewed_int'])
print(f"prior_offer_count - viewed 상관계수: {corr_viewed:.4f}, p-value: {p_viewed:.4g}, N={len(fatigue_df)}")

# --- 상관분석 2: prior_offer_count ↔ viewed_before_completed (bogo/discount, 수신한 인스턴스만) ---
comp_df = fatigue_df[fatigue_df['is_bogo_discount'] & fatigue_df['viewed']].copy()
comp_df['viewed_before_completed_int'] = comp_df['viewed_before_completed'].astype(int)
corr_comp, p_comp = pointbiserialr(comp_df['prior_offer_count'], comp_df['viewed_before_completed_int'])
print(f"prior_offer_count - viewed_before_completed 상관계수: {corr_comp:.4f}, p-value: {p_comp:.4g}, N={len(comp_df)}")
```

```python
# prior_offer_count 꼬리(표본 적은 큰 값들)는 묶어서 처리
MAX_COUNT = 5
fatigue_df['prior_offer_count_bin'] = fatigue_df['prior_offer_count'].clip(upper=MAX_COUNT)
comp_df['prior_offer_count_bin'] = comp_df['prior_offer_count'].clip(upper=MAX_COUNT)

viewed_summary = (
    fatigue_df.groupby('prior_offer_count_bin')['viewed_int']
    .agg(수신율='mean', n='size')
    .reset_index()
)
comp_summary = (
    comp_df.groupby('prior_offer_count_bin')['viewed_before_completed_int']
    .agg(완료율='mean', n='size')
    .reset_index()
)

xtick_labels = [str(i) if i < MAX_COUNT else f'{MAX_COUNT}+' for i in viewed_summary['prior_offer_count_bin']]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(viewed_summary['prior_offer_count_bin'], viewed_summary['수신율'] * 100,
        marker='o', color='#2a78d6', linewidth=2, label='수신율(viewed)')
ax.plot(comp_summary['prior_offer_count_bin'], comp_summary['완료율'] * 100,
        marker='o', color='#eb6834', linewidth=2, label='완료율(수신자 중, viewed_before_completed)')

for _, row in viewed_summary.iterrows():
    ax.text(row['prior_offer_count_bin'], row['수신율'] * 100 + 2, f"n={row['n']:,}",
            ha='center', fontsize=7.5, color='#52514e')

ax.set_xticks(viewed_summary['prior_offer_count_bin'])
ax.set_xticklabels(xtick_labels)
ax.set_xlabel('이전에 받은 오퍼 수 (prior_offer_count)')
ax.set_ylabel('비율(%)')
ax.set_title('오퍼 노출 순번에 따른 수신율·완료율 변화')
ax.set_ylim(0, 100)
ax.spines[['top', 'right']].set_visible(False)
ax.spines[['left', 'bottom']].set_color('#c3c2b7')
ax.yaxis.grid(True, color='#e1e0d9', linewidth=0.8, zorder=0)
ax.set_axisbelow(True)
ax.tick_params(colors='#898781')
ax.legend(frameon=False, loc='best')
plt.tight_layout()
plt.savefig('offer_fatigue.png', dpi=150)
plt.show()

print(viewed_summary)
print(comp_summary)
```

```python
# --- 2차 통제모델: offer_type + difficulty + reward + duration (8번 통합) ---
from statsmodels.stats.outliers_influence import variance_inflation_factor
model_8_controlled = smf.logit(
    'viewed_before_completed ~ C(offer_type) + difficulty + reward + duration',
    data=model_df
).fit()
print(model_8_controlled.summary())

# --- 다중공선성(VIF) 체크 — reward·offer_type 공선성이 실제로 얼마나 심한지 숫자로 확인 ---
X = pd.get_dummies(model_df[['offer_type', 'difficulty', 'reward', 'duration']], drop_first=True)
X = X.astype(float)
X.insert(0, 'const', 1.0)
vif_data = pd.DataFrame({
    'variable': X.columns,
    'VIF': [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
})
print(vif_data)
```

```python
# 11. Reward 금액이 고소득자의 동기부여 수준에 미달
# --- 소득 사분위 그룹 (10번에서 이미 만들었으면 재사용 가능) ---
from scipy.stats import chi2_contingency
model_df_income['income_q'] = pd.qcut(model_df_income['income'], 4, labels=['Q1(저)', 'Q2', 'Q3', 'Q4(고)'])

# --- 소득 분위별로 reward-완료 카이제곱 검정 (분위마다 reward 반응성이 다른지) ---
for q in model_df_income['income_q'].cat.categories:
    sub = model_df_income[model_df_income['income_q'] == q]
    contingency = pd.crosstab(sub['reward'], sub['viewed_before_completed'])
    chi2, p, dof, expected = chi2_contingency(contingency)
    print(f"{q}: chi2={chi2:.2f}, p={p:.4g}, N={len(sub)}")

# --- 집단별 완료율 표 (income 분위 x reward) ---
summary_11 = (
    model_df_income.groupby(['income_q', 'reward'])['viewed_before_completed']
    .agg(완료율='mean', n='size')
    .reset_index()
)
print(summary_11)

# --- 로지스틱 회귀: income x reward 상호작용 ---
model_11 = smf.logit('viewed_before_completed ~ income * reward', data=model_df_income).fit()
print(model_11.summary())
```

```python
coef_df = pd.DataFrame({
    'coef': model_B.params,
    'std_err': model_B.bse,
    'z': model_B.tvalues,
    'p_value': model_B.pvalues
})
print(coef_df)
print("pseudo R-squ.:", model_B.prsquared)
print("N:", int(model_B.nobs))
```

```python
completed_events = transcript[
    (transcript['event'] == 'offer completed') &
    (transcript['offer_id'].isin(bogo_discount_ids))
].copy()

viewed_events = transcript[
    (transcript['event'] == 'offer viewed') &
    (transcript['offer_id'].isin(bogo_discount_ids))
]

viewed_lookup = viewed_events.groupby(['person', 'offer_id'])['time'].apply(list).to_dict()

def has_prior_view(row):
    view_times = viewed_lookup.get((row['person'], row['offer_id']), [])
    return any(vt <= row['time'] for vt in view_times)

completed_events['viewed_before'] = completed_events.apply(has_prior_view, axis=1)

no_view_completed = completed_events[~completed_events['viewed_before']]

print("전체 완료 이벤트 수:", len(completed_events))
print("열람 없이 완료된 건수:", len(no_view_completed))
print("비율(%):", round(len(no_view_completed) / len(completed_events) * 100, 2))
```
