```python
!pip install python-pptx
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.dml.color import RGBColor
from pptx.enum.text import PP_ALIGN
from pptx.enum.shapes import MSO_SHAPE

# 1. 프레젠테이션 객체 생성
prs = Presentation()
prs.slide_width = Inches(13.333)  # 16:9 와이드스클린
prs.slide_height = Inches(7.5)

# 브랜드 컬러 정의
COLOR_DARK_GREEN = RGBColor(0, 98, 65)      # #006241 (스타벅스 시그니처 그린)
COLOR_OFF_WHITE = RGBColor(243, 243, 243)    # #F3F3F3 (배경)
COLOR_WHITE = RGBColor(255, 255, 255)        # #FFFFFF (카드 배경)
COLOR_GOLD = RGBColor(203, 162, 88)         # #CBA258 (포인트 골드)
COLOR_CHARCOAL = RGBColor(33, 33, 33)        # #212121 (본문 텍스트)
COLOR_LIGHT_GREEN = RGBColor(212, 233, 226)  # #D4E9E2 (포인트 배경)

blank_slide_layout = prs.slide_layouts[6]

# ================= =================
# [슬라이드 1] 표지 (Title Slide)
# ================= =================
slide1 = prs.slides.add_slide(blank_slide_layout)

# 다크 그린 전체 배경
bg1 = slide1.shapes.add_shape(MSO_SHAPE.RECTANGLE, 0, 0, prs.slide_width, prs.slide_height)
bg1.fill.solid()
bg1.fill.fore_color.rgb = COLOR_DARK_GREEN
bg1.line.fill.background()

# 타이틀 텍스트 상자
tx_box = slide1.shapes.add_textbox(Inches(1.0), Inches(2.2), Inches(11.333), Inches(3.5))
tf = tx_box.text_frame
tf.word_wrap = True

p0 = tf.paragraphs[0]
p0.text = "Starbucks Offer Analytics"
p0.font.name = "Noto Sans KR"
p0.font.size = Pt(20)
p0.font.bold = True
p0.font.color.rgb = COLOR_GOLD
p0.space_after = Pt(14)

p1 = tf.add_paragraph()
p1.text = "오퍼 구조 및 고객 프로필 기반\n완료율 저하 요인 분석"
p1.font.name = "Noto Sans KR"
p1.font.size = Pt(40)
p1.font.bold = True
p1.font.color.rgb = COLOR_OFF_WHITE
p1.space_after = Pt(20)

p2 = tf.add_paragraph()
p2.text = "Logic Tree & Statistical Verification Report"
p2.font.name = "Noto Sans KR"
p2.font.size = Pt(16)
p2.font.color.rgb = COLOR_LIGHT_GREEN


# ================= =================
# [슬라이드 2] 본문 - 로직 트리 요약 (Content Slide)
# ================= =================
slide2 = prs.slides.add_slide(blank_slide_layout)

# 연한 오프화이트 배경
bg2 = slide2.shapes.add_shape(MSO_SHAPE.RECTANGLE, 0, 0, prs.slide_width, prs.slide_height)
bg2.fill.solid()
bg2.fill.fore_color.rgb = COLOR_OFF_WHITE
bg2.line.fill.background()

# 상단 헤더
header_box = slide2.shapes.add_textbox(Inches(0.8), Inches(0.6), Inches(11.7), Inches(1.0))
htf = header_box.text_frame
hp0 = htf.paragraphs[0]
hp0.text = "01. Executive Summary: 원인 분해 프레임워크"
hp0.font.name = "Noto Sans KR"
hp0.font.size = Pt(26)
hp0.font.bold = True
hp0.font.color.rgb = COLOR_DARK_GREEN

# 3개의 주요 카드 배치
card_data = [
    ("구매 동기 및 수용성", "Motivation & Fit", ["매력도 부족: 소비 부담 대비 낮은 Reward", "구매 관성: 기존 결제/메뉴 패턴 고착화", "호감도 불일치: 개인 선호와 오퍼 불일치🔴"]),
    ("인지 및 접촉 제약", "Awareness & Reach", ["노출 채널 한계: Web/Email 사용률 저하", "기간 부족: Duration 미달 전 마감", "노출 둔화: 반복 피로도 증가"]),
    ("달성 난이도 부담", "Barrier & Difficulty", ["소비 금액 증가: Difficulty 도달 부담", "과잉 구매 부담: BOGO 등 수량 유인 부담", "소득 연계: 소득 대비 상대적 난이도 장벽"])
]

card_width = Inches(3.64)
card_height = Inches(4.8)
card_gap = Inches(0.4)
left_margin = Inches(0.8)

for i, (title_kr, title_en, bullets) in enumerate(card_data):
    x_pos = left_margin + i * (card_width + card_gap)
    y_pos = Inches(1.8)
    
    # 카드 배경 (둥근 사각형)
    card = slide2.shapes.add_shape(MSO_SHAPE.ROUNDED_RECTANGLE, x_pos, y_pos, card_width, card_height)
    card.fill.solid()
    card.fill.fore_color.rgb = COLOR_WHITE
    card.line.color.rgb = COLOR_DARK_GREEN
    card.line.width = Pt(1.5)
    
    # 카드 텍스트
    ctf = card.text_frame
    ctf.word_wrap = True
    ctf.margin_top = Inches(0.3)
    ctf.margin_left = Inches(0.3)
    ctf.margin_right = Inches(0.3)
    
    cp0 = ctf.paragraphs[0]
    cp0.text = title_kr
    cp0.font.name = "Noto Sans KR"
    cp0.font.size = Pt(20)
    cp0.font.bold = True
    cp0.font.color.rgb = COLOR_DARK_GREEN
    
    cp1 = ctf.add_paragraph()
    cp1.text = title_en
    cp1.font.name = "Noto Sans KR"
    cp1.font.size = Pt(12)
    cp1.font.color.rgb = COLOR_GOLD
    cp1.space_after = Pt(20)
    
    for bullet in bullets:
        bp = ctf.add_paragraph()
        bp.text = f"• {bullet}"
        bp.font.name = "Noto Sans KR"
        bp.font.size = Pt(13)
        bp.font.color.rgb = COLOR_CHARCOAL
        bp.space_after = Pt(10)

# 3. PPT 파일 저장
prs.save("Starbucks_Offer_Analytics_Template.pptx")
print("✅ 스타벅스 테마 PPTX 파일이成功적으로 생성되었습니다: Starbucks_Offer_Analytics_Template.pptx")
```
