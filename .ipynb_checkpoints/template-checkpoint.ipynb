```python
#스케일링
import os, joblib
from sklearn.preprocessing import RobustScaler

def 스케일링함수(data, scaler_path="scaler.pkl"):
    data = data.copy()
    if os.path.exists(scaler_path):
        scaler = joblib.load(scaler_path)
        data["컬럼_scaled"] = scaler.transform(data[["컬럼"]])
    else:
        scaler = RobustScaler()
        data["컬럼_scaled"] = scaler.fit_transform(data[["컬럼"]])
        joblib.dump(scaler, scaler_path)
    return data
```

```python
#인코딩
# One-Hot (명목형, 3개 이상)
dummies = pd.get_dummies(df["컬럼"], prefix="컬럼", dtype=int)
df = pd.concat([df.drop(columns=["컬럼"]), dummies], axis=1)

# Label (고정 매핑, 순서/2개짜리)
df["컬럼"] = df["컬럼"].map({"값1": 0, "값2": 1})
```

```python
#groupby로 최고/최저 찾기
result = df.groupby("기준컬럼")["값컬럼"].sum().sort_values(ascending=False)
top_name = result.idxmax()
top_value = result.max()
```
